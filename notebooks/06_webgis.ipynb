# Aset spasial yang tersedia

Aset referensi dari PDF kini disimpan terpisah di `data/knn_pdf_reference/`:

- `hasil_klasifikasi_knn.html` - peta regional KNN siap dibuka di browser.
- `jatim_250m.tif` - raster regional yang disebut laporan.
- `jawa_timur_provinsi.geojson` dan enam ZIP sampel kelas - batas wilayah dan sampel.
- `prediksi_data_uji.csv` - ringkasan hasil evaluasi 268 poligon.

Peta HTML merupakan hasil yang sudah diproses dari repo sumber. Dataset, raster, dan batas sampel juga tersedia sebagai bahan rujukan; file model dan pipeline KNN untuk membangun ulang peta tidak disediakan.
