# BAB 1 - Business Understanding

# Klasifikasi spasial tutupan lahan Jawa Timur

Proyek ini mengotomatisasi pemetaan enam kelas tutupan lahan menggunakan citra Sentinel-2A Level-2A dan machine learning. Tujuan utamanya adalah menghasilkan prediksi kelas piksel dan menyajikan hasilnya dalam peta evaluasi poligon serta peta inferensi regional.

## Kelas target

| Kode | Kelas |
|---:|---|
| 1 | Sawah |
| 2 | Bangunan |
| 3 | Hutan (lahan hijau) |
| 4 | Danau |
| 5 | Laut (perairan terbuka) |
| 6 | Mangrove |

## Pendekatan

Laporan menggunakan K-Nearest Neighbors (KNN) atas fitur band dan indeks spektral yang telah distandarisasi. Pemilihan KNN ditujukan untuk menangkap batas keputusan non-linear pada data reflektansi. Detail sumber data dan keterbatasan evaluasi dijelaskan pada bab berikutnya.
