# K-Nearest Neighbors

PDF melaporkan `KNeighborsClassifier` dengan konfigurasi berikut:

| Parameter | Nilai |
|---|---|
| Jumlah tetangga | 5 |
| Metrik jarak | Euclidean |
| Pembobotan suara | Uniform |
| Penskalaan | `StandardScaler` |

KNN memprediksi kelas piksel berdasarkan mayoritas kelas pada lima tetangga terdekat dalam ruang fitur terstandarisasi. Penskalaan diperlukan karena nilai reflektansi band dan indeks memiliki rentang berbeda.

## Fitur dan artefak

Dataset berisi lima band optik (B02, B03, B04, B08, B11) dan empat indeks (NDVI, NDWI, MNDWI, NDBI), total sembilan fitur. Peta regional KNN yang sudah dihasilkan tersedia sebagai `data/knn_pdf_reference/hasil_klasifikasi_knn.html`.

File model KNN terlatih dan pipeline training KNN tidak tersedia di repository sumber. Notebook sumber yang tersedia di repo tersebut membahas eksperimen Random Forest, bukan konfigurasi KNN pada PDF. Karena itu, bab ini mendokumentasikan konfigurasi PDF tanpa mengklaim telah melatih atau mengekspor model.
