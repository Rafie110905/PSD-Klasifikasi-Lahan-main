# Sumber dan karakteristik data

Aset data yang dirujuk PDF kini tersedia di `data/knn_pdf_reference/`. Sumbernya adalah repository publik `itsaalii/PSD`, commit `687183bddc9f78a89ef701642bde9a634a80f63c`.

## Dataset dan input spasial

- `dataset.csv`: 322.146 baris piksel, 19 kolom, 274 ID poligon, dan enam kelas.
- `jatim_250m.tif`: raster regional yang disebut dalam laporan untuk inferensi.
- `jawa_timur_provinsi.geojson`: batas provinsi.
- Enam arsip ZIP sampel poligon: Sawah, Bangunan, Hutan, Danau, Laut, dan Mangrove.
- Dataset menyimpan koordinat, label, lima band optik (B02, B03, B04, B08, B11), dan empat indeks (NDVI, NDWI, MNDWI, NDBI).

| Fitur | Makna umum |
|---|---|
| B02 (Blue), B03 (Green), B04 (Red) | Pantulan spektrum tampak; membantu membedakan air, vegetasi, dan permukaan terbangun |
| B08 (NIR) | Pantulan vegetasi dan serapan air |
| B11 (SWIR) | Kelembapan permukaan dan material terbangun |
| NDVI | Kerapatan vegetasi |
| NDWI, MNDWI | Keberadaan badan air |
| NDBI | Kawasan terbangun |

Jumlah baris, kolom, kelas, dan poligon telah diperiksa dari CSV yang diunduh. Nilai evaluasi laporan tersedia terpisah; rincian asal dan batasan aset dijelaskan dalam `data/knn_pdf_reference/README.md`.
