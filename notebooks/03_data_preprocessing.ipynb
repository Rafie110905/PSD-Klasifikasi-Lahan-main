# Penyiapan fitur dan pembagian data

Dataset sumber tersedia di `data/knn_pdf_reference/dataset.csv`. Sesuai PDF, alurnya menyaring awan menggunakan SCL, mengekstrak lima band dan empat indeks spektral pada piksel sampel, lalu membagi data secara stratified dengan rasio 80:20.

## Pembagian yang dilaporkan

| Bagian | Jumlah piksel |
|---|---:|
| Latih | 257.716 |
| Uji | 64.430 |
| Total | 322.146 |

Laporan menyebut `random_state=42` dan stratifikasi berdasarkan label. Fitur distandarisasi memakai `StandardScaler`; parameter scaler dipelajari dari data latih dan diterapkan pada data uji.

## Catatan evaluasi spasial

Split dilakukan pada tingkat piksel, bukan dengan memisahkan seluruh poligon. Akibatnya, piksel dari poligon yang sama dapat terbagi ke data latih dan uji. Evaluasi seperti ini tidak setara dengan pengujian generalisasi pada poligon atau lokasi yang sama sekali belum pernah dilihat.
